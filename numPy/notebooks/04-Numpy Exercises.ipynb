#### Create the following matrix:

```python
mat = np.ones((10,10))
mat = mat / 100
mat_length = mat.shape[1]
for i in range(mat_length):
    if i == 0:
        mat[i] = np.arange(1,11) / 100
    else:
        mat[i] = mat[i-1]+0.1
mat
```

### Now do the following

#### Get the sum of all the values in mat

```python
sum = mat[0] + mat[1] + mat[2] + mat[3] + mat[4]
sum = sum[0] + sum[1] + sum[2] + sum[3] + sum[4]
print(sum)
```

```python
sum = 0
total_sum = 0
for m in mat:
    sum += m
for s in sum:
    total_sum += s

print(total_sum)
```

```python
print(mat.sum())
```

#### Get the standard deviation of the values in mat

```python
print(np.std(mat))
```

#### Get the sum of all the columns in mat

```python
sum = 0
for m in mat:
    sum += m
print(sum)
```

```python
print(mat.sum(axis=0))
```
